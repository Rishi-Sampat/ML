```python
def convert_to_gb(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip().upper()

    if value.endswith("TB"):
        return float(value.replace("TB", "")) * 1024

    elif value.endswith("GB"):
        return float(value.replace("GB", ""))

    return np.nan

dt['storage'] = dt['storage'].apply(convert_to_gb)
dt['ssd_size'] = dt['ssd_size'].apply(convert_to_gb)
dt['graphics_card_size'] = dt['graphics_card_size'].apply(convert_to_gb)
```
