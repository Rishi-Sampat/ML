```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import convolve
```

```python
img=cv2.imread("/content/imagesssss.jpg",cv2.IMREAD_GRAYSCALE)
```

```python
plt.figure(figsize=(6,4))
plt.imshow(img,cmap='gray')
```

```python
sharpen=[[0,-1,0],[-1,5,-1],[0,-1,0]] #depends on the intensity of image, according to the image , if we have
#lighter image = positive weight and negative values
#darker image = negative weight and positive values
```

```python
kernel=sharpen
#perform the convolution operation

im=np.array(img,dtype=float)
img2=convolve(im,kernel)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.imshow(img,cmap='gray')
plt.subplot(1,2,2)
plt.imshow(img2,cmap='gray')
```

```python
sharpen2=[[0,1,0],[1,-5,1],[0,1,0]]
```

```python
kernel2=sharpen2
im2=np.array(img,dtype=float)
img3=convolve(im2,kernel2)
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(img,cmap='gray')
plt.subplot(1,3,2)
plt.imshow(img2,cmap='gray')
plt.subplot(1,3,3)
plt.imshow(img3,cmap='gray')
```
